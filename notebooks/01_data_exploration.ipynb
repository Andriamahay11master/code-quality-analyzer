```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#create function for load data
def load_data(path, encoding="utf-8"):
  data = pd.read_csv(path, encoding=encoding)
  return data

#function for preview data, checking shape, missing values, columns list
def preview(df, name="DataFrame"):
  print(f"\n{name} shape: {df.shape}")
  print(df.head())
  print("\nMissing values:")
  print(df.isnull().sum())
  print("\nColumns:")
  print(df.columns.tolist())

```

```python
path_repo = "/content/drive/MyDrive/Analyzer-code-data/repositories.csv"
path_attribute = "/content/drive/MyDrive/Analyzer-code-data/attribute-details.csv"
path_versions = "/content/drive/MyDrive/Analyzer-code-data/versions.csv"
path_codesmells = "/content/drive/MyDrive/Analyzer-code-data/codesmells.csv"
```

```python
#Load the data
data_repo = load_data(path_repo)
data_attribute = load_data(path_attribute)
data_versions = load_data(path_versions)
data_codesmells = load_data(path_codesmells, "cp1252")
```

```python
#Preview the data
preview(data_repo, "data_repo")
preview(data_attribute, "data_attribute")
preview(data_versions, "data_versions")
preview(data_codesmells, "data_codesmells")
```
